```python
#Exercise 1.1: Mach Bands
import numpy as np
import matplotlib.pyplot as plt

bands = np.zeros((100, 600))
for i in range(6):
    bands[:, i*100:(i+1)*100] = i * 40

plt.imshow(bands, cmap='gray')
plt.title("Synthetic Mach Bands")
plt.axis('off')
plt.show()
```

```python
#Exercise 2.1: Simultaneous Contrast
img = np.ones((200, 600)) * 100
img[:, :200] = 20
img[:, 400:] = 200
img[50:150, 250:350] = 120

plt.imshow(img, cmap='gray')
plt.title("Simultaneous Contrast Example")
plt.axis('off')
plt.show()
```

```python
#Exercise 3.1: Image as Matrix
image = np.random.randint(0, 256, (256, 256))

print("Shape:", image.shape)
print("Min:", image.min())
print("Max:", image.max())
```

```python
#Exercise 4.1: Sampling
original = image.copy()

down_64 = original[::4, ::4]
down_32 = original[::8, ::8]

plt.figure(figsize=(9,3))

plt.subplot(1,3,1)
plt.imshow(original, cmap='gray')
plt.title("256x256")

plt.subplot(1,3,2)
plt.imshow(down_64, cmap='gray')
plt.title("64x64")

plt.subplot(1,3,3)
plt.imshow(down_32, cmap='gray')
plt.title("32x32")

plt.show()
```

```python
#Exercise 5.1: Quantisation
def quantise(img, levels):
    step = 256 // levels
    return (img // step) * step

q256 = quantise(image, 256)
q16 = quantise(image, 16)
q4 = quantise(image, 4)

plt.figure(figsize=(9,3))

plt.subplot(1,3,1)
plt.imshow(q256, cmap='gray')
plt.title("256 levels")

plt.subplot(1,3,2)
plt.imshow(q16, cmap='gray')
plt.title("16 levels")

plt.subplot(1,3,3)
plt.imshow(q4, cmap='gray')
plt.title("4 levels")

plt.show()
```

```python
#Exercise 8.1: Distance
import math

p = (2,3)
q = (7,9)

D_euclid = math.sqrt((p[0]-q[0])**2 + (p[1]-q[1])**2)
D4 = abs(p[0]-q[0]) + abs(p[1]-q[1])
D8 = max(abs(p[0]-q[0]), abs(p[1]-q[1]))

print(D_euclid, D4, D8)
```

```python
#Exercise 9.1: Translation
translated = np.zeros_like(image)
translated[20:, 30:] = image[:-20, :-30]

plt.imshow(translated, cmap='gray')
plt.title("Translated Image")
plt.axis('off')
plt.show()
```

```python
#Exercise 10.1: Nearest Neighbor Zoom
down_32 = image[::8, ::8]

zoom = np.repeat(np.repeat(down_32, 8, axis=0), 8, axis=1)

plt.imshow(zoom, cmap='gray')
plt.title("Nearest Neighbor Zoom")
plt.axis('off')
plt.show()
```

```python
#Exercise 11: Mini Project
from skimage import data

img = data.camera()

# Spatial resolution
low1 = img[::2, ::2]
low2 = img[::4, ::4]
low3 = img[::8, ::8]

plt.figure(figsize=(10,3))

plt.subplot(1,3,1)
plt.imshow(low1, cmap='gray')
plt.title("Low Res 1")

plt.subplot(1,3,2)
plt.imshow(low2, cmap='gray')
plt.title("Low Res 2")

plt.subplot(1,3,3)
plt.imshow(low3, cmap='gray')
plt.title("Low Res 3")

plt.show()

# Intensity resolution
def quantise(img, levels):
    step = 256 // levels
    return (img // step) * step

q1 = quantise(img, 64)
q2 = quantise(img, 16)
q3 = quantise(img, 4)

plt.figure(figsize=(10,3))

plt.subplot(1,3,1)
plt.imshow(q1, cmap='gray')
plt.title("64 levels")

plt.subplot(1,3,2)
plt.imshow(q2, cmap='gray')
plt.title("16 levels")

plt.subplot(1,3,3)
plt.imshow(q3, cmap='gray')
plt.title("4 levels")

plt.show()
```
